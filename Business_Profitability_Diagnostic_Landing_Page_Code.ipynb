# Business Profitability Diagnostic — HTML Source Code

## Project Overview

This notebook documents the HTML source for the landing page of the **Business Profitability Diagnostic** application. The interface is written in Spanish for its intended users; the documentation in this notebook is in English for portfolio presentation.

The page introduces a profitability diagnostic for small businesses and independent professionals. It explains the value proposition, previews a simple revenue-versus-cost example, describes the user journey, and directs visitors to the diagnostic page.

> **Scope note:** This file is the landing page, not the interactive diagnostic form itself. Both primary calls to action link to `diagnostico-rentabilidad.html`, which must be available in the same deployed website path.

## Tools and Techniques

- **HTML5:** semantic page structure, headings, lists, links, and accessible labels.
- **CSS:** responsive layout, visual design, reusable custom properties, cards, and call-to-action styling.
- **JavaScript:** light/dark theme toggle, browser color-scheme detection, and preference persistence using `localStorage`.
- **Accessibility considerations:** descriptive labels, visible keyboard focus styles, and reduced-motion preferences.

## Code Structure

| Section | Purpose |
|---|---|
| Document metadata | Sets the language, encoding, responsive viewport, page title, description, and font. |
| CSS design system | Defines color variables, typography, responsive layouts, cards, buttons, and motion behavior. |
| Page body | Renders the hero banner, example margin bar, feature cards, process steps, and closing call to action. |
| JavaScript | Restores and toggles the selected color theme. |

The HTML source below has been formatted and commented for readability. Its visible Spanish interface text and core behavior have been preserved. The HTML is shown as source code for review; it is not executed as a Python cell.

## Implementation Notes and Review Checklist

- **Navigation dependency:** the “Hacer mi diagnóstico” links point to `diagnostico-rentabilidad.html`. Verify that this file exists alongside this landing page in the deployed site.
- **Theme behavior:** the page first tries to restore the theme from `localStorage`; the toggle then switches between light and dark mode and saves the selection when browser storage is available.
- **Responsive behavior:** CSS grid breakpoints adapt the feature cards and process steps to wider screens, while the bar text adjusts for narrow screens.
- **External dependency:** the page loads the Plus Jakarta Sans font from Google Fonts. The page falls back to system fonts if the external font is unavailable.
- **Content language:** interface labels and explanatory text are intentionally in Spanish; the portfolio documentation is in English.
- **Scope limitation:** this landing-page file does not contain the actual calculation workflow. That is linked as a separate HTML page and should be reviewed separately before claiming its detailed financial logic in a portfolio description.
